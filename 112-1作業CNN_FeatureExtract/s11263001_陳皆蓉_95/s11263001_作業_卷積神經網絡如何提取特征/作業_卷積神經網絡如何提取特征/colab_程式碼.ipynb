```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def apply_convolution(image, kernel):
    return cv2.filter2D(image, -1, kernel)

# 讀取原始圖像，並轉換成灰階圖
image = cv2.imread('house.jpg', cv2.IMREAD_GRAYSCALE)

# 定義直線、橫線和斜線卷積核
kernel_vertical = np.array([[-1, 0, 1],
               [-1, 0, 1],
               [-1, 0, 1]])

kernel_horizontal = np.array([[-1, -1, -1],
                [0, 0, 0],
                [1, 1, 1]])

kernel_diagonal = np.array([[1, 0, -1],
               [0, 0, 0],
               [-1, 0, 1]])

# 應用卷積核並獲取特征圖
vertical_feature = apply_convolution(image, kernel_vertical)
horizontal_feature = apply_convolution(image, kernel_horizontal)
diagonal_feature = apply_convolution(image, kernel_diagonal)

# 顯示原圖和特征圖
plt.figure(figsize=(10, 5))

plt.subplot(2, 2, 1)
plt.imshow(image, cmap='gray')
plt.title('Original Image')

plt.subplot(2, 2, 2)
plt.imshow(vertical_feature, cmap='gray')
plt.title('Vertical Feature')

plt.subplot(2, 2, 3)
plt.imshow(horizontal_feature, cmap='gray')
plt.title('Horizontal Feature')

plt.subplot(2, 2, 4)
plt.imshow(diagonal_feature, cmap='gray')
plt.title('Diagonal Feature')

plt.show()

# 保存特徵圖到Colab的文件夾
cv2.imwrite('/content/vertical_feature.jpg', vertical_feature)
cv2.imwrite('/content/horizontal_feature.jpg', horizontal_feature)
cv2.imwrite('/content/diagonal_feature.jpg', diagonal_feature)
```
