```python
import numpy as np
import cv2
from google.colab.patches import cv2_imshow
import urllib


# 載入網路圖像
req = urllib.request.urlopen('https://www.shutterstock.com/image-photo/london-england-august-2016-full-260nw-668643088.jpg')
arr = np.asarray(bytearray(req.read()), dtype=np.uint8)
img = cv2.imdecode(arr, -1) # 'Load it as it is'

# 定義直線卷積核
kernel_v = np.array([[1, 0, -1],
 [2, 0, -2],
 [1, 0, -1]])

# 定義橫線卷積核
kernel_h = np.array([[1, 2, 1], [0, 0, 0], [-1, -2, -1]])

# 定義斜線卷積核
kernel_s = np.array([[1, 0, -1], [0, 0, 0], [-1, 0, 1]])

# img = cv2.imread("pikachu.jpg")

# 轉換圖像為灰度圖像
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# 進行卷積操作
filtered_h = cv2.filter2D(gray_img, -1, kernel_h)
filtered_v = cv2.filter2D(gray_img, -1, kernel_v)
filtered_s = cv2.filter2D(gray_img, -1, kernel_s)

# 顯示輸出圖像
print("原圖")
cv2_imshow(img)
print("直線卷積")
cv2_imshow(filtered_v)
print("橫線卷積")
cv2_imshow(filtered_h)
print("斜線卷積")
cv2_imshow(filtered_s)
cv2.waitKey(0)
```
